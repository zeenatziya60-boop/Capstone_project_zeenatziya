```python
# Making a Google Translator using Gemini

from google.genai import types, errors

def translate_text(text_to_translate, target_language):

  try:
    response = client.models.generate_content(
      model="gemini-3-flash-preview",
      contents=text_to_translate,
      config=types.GenerateContentConfig(
        system_instruction=f"Translate this into {target_language}. Keep the same feeling. Reply with only the text."
      )
    )
    return {"status":"Success",
      "data": response.text,
      "tokens_used": response.usage_metadata.total_token_count}

  except errors.APIError as e:
    return {"status":"Error","data":e.message,"tokens_used":0}

```

```python
def generate_notes(topic, creative=False):

    temp = 1.2 if creative else 0.2

    try:

        response = client.models.generate_content(
            model="gemini-3.1-flash-lite",
            contents=f"Create notes on {topic}.",
            config=types.GenerateContentConfig(
                temperature=temp,
                max_output_tokens=1000,
                # stop_sequences=["###"],
                http_options=types.HttpOptions(
                    timeout=15000
                )
            )
        )

        return {
            "notes": response.text.strip(),
            "finish_reason": response.candidates[0].finish_reason,
            "tokens": response.usage_metadata.total_token_count
        }

    except Exception:
        return {
            "notes": "Request Failed",
            "finish_reason": "ERROR",
            "tokens": 0
        }
```

```python
result = generate_notes(
    "AI in Cricket",
    creative=False
)

print(result)
```

Imagine a cloud company gets messy, plain-English complaints from customers all day: "the site is down," "everything is so slow," "I can't log in." **Our tool needs to read a raw complaint, and automatically:**

- Work out which system is likely affected
- Judge how severe the issue is
- Recommend an action
- Package it all cleanly — and never crash, even if the network misbehaves



1.   "Our database is not responding and users are seeing 500 Internal Server Error."
2.   "Customers cannot access the website because the network connection keeps dropping."
3.   "The login page loads, but clicking the Sign In button does nothing."



```python
from google import genai
from google.colab import userdata
from google.genai import types, errors

key = userdata.get('SY_API_KEY')
client = genai.Client(api_key = key)

# Incident triage is the rapid assessment step where IT and operations teams determine the scope, severity, and ownership of a new incident before active resolution begins.

def triage_incident(raw_complaint):

  #detailed system instruction which has persona, task and output format
  system_blueprint = """
    You are an automated cloud server triage engine.
    Extract data into this exact schema:
    System: [Database, Frontend, Network, or Unknown]
    Severity: [High, Medium, Low]
    Action: [Restart, Scale, Patch, or Investigate]

    EXAMPLE:
    Input: "Our database connection pool is exhausted! 500 errors."
    Output: System: Database | Severity: High | Action: Restart
    """

  try:
    response = client.models.generate_content(

            model="gemini-3.1-flash-lite",
            contents=f"Alert Text: <{raw_complaint}>",

            #parameter tuning
            config=types.GenerateContentConfig(
                system_instruction=system_blueprint,
                temperature=0.0,
                max_output_tokens=250,
                http_options=types.HttpOptions(timeout=15000)
            )
        )
    #removing extra spaces
    raw_output = response.text.strip()

    #final output when the code runs successfully
    return {"status": "Success", "payload": raw_output,
                "tokens": response.usage_metadata.total_token_count}

  except errors.APIError as e:
      return {"status": "Pipeline Error", "payload": str(e.message), "tokens": 0}

  except Exception as e:
      return {"status": "Pipeline Error", "payload": str(e), "tokens": 0}


```

```python
result = triage_incident("The login page loads, but clicking the Sign In button does nothing")

print(result)
```

```python
import time

def resilient_triage_orchestrator(complaint_text, max_retries=3):
    print("--- STARTING TRIAGE ---")

    for attempt in range(1, max_retries + 1):
        print(f"Attempt {attempt} of {max_retries}...")
        report = triage_incident(complaint_text)

        if report["status"] == "Success":
            print("Success!")
            return report

        print(f"Attempt {attempt} failed. Retrying in 2 seconds...")
        time.sleep(2)

    print("All retries used up without success.")
    return report
```
